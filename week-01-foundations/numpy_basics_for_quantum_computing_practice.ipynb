## 13. Quantum Gates as Matrices


Quantum gates are matrices that transform quantum state vectors.

Common single-qubit gates:

- I: identity
- X: bit flip
- Z: phase flip
- H: creates superposition

```python
I = np.eye(2)

X = np.array([
    [0, 1],
    [1, 0]
])

Z = np.array([
    [1, 0],
    [0, -1]
])

H = (1 / np.sqrt(2)) * np.array([
    [1, 1],
    [1, -1]
])

print("I:\n", I)
print("X:\n", X)
print("Z:\n", Z)
print("H:\n", H)
```

## 14. Probabilities from Amplitudes


Quantum states contain amplitudes.

## 16. Tensor Product using `np.kron()`


`np.kron()` is used to combine quantum states and gates.

This is essential for multi-qubit quantum computing.

Examples:

- `|00⟩ = np.kron(|0⟩, |0⟩)`
- `|01⟩ = np.kron(|0⟩, |1⟩)`

### Example

```python
zero = np.array([1, 0])
one = np.array([0, 1])

zero_zero = np.kron(zero, zero)
zero_one = np.kron(zero, one)

print("|00>:", zero_zero)
print("|01>:", zero_one)
```

# Final Mini Project: 1-Qubit Quantum Simulator

Build a small simulator with these features:

1. Create `|0⟩` and `|1⟩`.
2. Create I, X, Z, and H gates.
3. Apply gates using `@`.
4. Print final state.
5. Print measurement probabilities.
6. Add helper functions:
   - `apply_gate(gate, state)`
   - `get_probabilities(state)`
   - `normalize(state)`

```python
import numpy as np

def apply_gate(gate, state):
    return gate @ state

def get_probabilities(state):
    return np.abs(state) ** 2

def normalize(state):
    return state / np.linalg.norm(state)

zero = np.array([1, 0])
one = np.array([0, 1])

I = np.eye(2)

X = np.array([
    [0, 1],
    [1, 0]
])

Z = np.array([
    [1, 0],
    [0, -1]
])

H = (1 / np.sqrt(2)) * np.array([
    [1, 1],
    [1, -1]
])

state = zero

state = apply_gate(H, state)
print("After H:", state)
print("Probabilities:", get_probabilities(state))

state = apply_gate(X, state)
print("After X:", state)
print("Probabilities:", get_probabilities(state))
```
